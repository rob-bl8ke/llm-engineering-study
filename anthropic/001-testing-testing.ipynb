```python
import os

# Load env variables
from dotenv import load_dotenv
load_dotenv()

api_key = os.getenv('ANTHROPIC_API_KEY')
print(api_key)
```

So now we start looking at the Claude API. Note how we must pass the `api_key` in and also, if the model is deprecated, it won't work.

```python
from anthropic import Anthropic

client = Anthropic(api_key=api_key)
model = "claude-haiku-5-5"
```

```python
message = client.messages.create(model=model, max_tokens=1000, messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence."
        }
    ]
)
```
